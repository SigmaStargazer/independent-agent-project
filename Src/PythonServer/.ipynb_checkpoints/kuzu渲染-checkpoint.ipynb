```python
db_path = "db/graphiti.kuzu"
```

```python
# cypher = "MATCH (a)-[b]->(c) RETURN * LIMIT 100"
cypher = """MATCH (a)-[b]->(c)
WHERE a.group_id = 'e5b08fe6988e'
RETURN * LIMIT 100"""
```

```python
try:
    # 1. 以只读模式打开 (如果 Kuzu 版本支持) 或者普通模式
    db = kuzu.Database(db_path, read_only=True)
    conn = kuzu.Connection(db)
    
    # 2. 画图
    print("🎨 正在加载最新记忆图谱...")
    g = KuzuGraphWidget(conn)
    display(g)
    
    # 3. 查询
    g.show_cypher(cypher)

except Exception as e:
    print(f"⚠️ 无法读取数据库，可能智能体正在写入: {e}")
```

```python
import gc

# 1. 尝试关闭连接 (如果有 close 方法)
try:
    if 'conn' in locals() and conn:
        del conn
    if 'db' in locals() and db:
        del db
    if 'g' in locals() and g:
        del g  # 组件也必须删掉，它也持有引用
except:
    pass

# 2. 强制垃圾回收 (关键)
gc.collect()
```
