```python
import re
```

```python
# 正则表达式
pattern = r'^\d+[smh]$'
```

```python
# 自定义异常类
class InvalidFormatError(Exception):
    def __init__(self, message="Invalid format. String must be a number followed by 's', 'm', or 'h'."):
        self.message = message
        super().__init__(self.message)

# 校验函数
def validate_string(s):
    if not re.match(pattern, s):
        raise InvalidFormatError()
    return True
```

```python
# 测试代码
test_strings = ["10s", "5m", "2h", "10", "10x", "s", "10s10"]

for test in test_strings:
    try:
        validate_string(test)
        print(f"{test}: Valid")
    except InvalidFormatError as e:
        print(f"{test}: {e}")
```
